```python
import subprocess

policy = "HuggingFaceVLA/smolvla_libero"
task_ids = [0, 1, 2]
episodes_per_task = 5  # Pilot; increase if your Colab time allows.
output_root = "/content/drive/MyDrive/smolvla_libero_pilot"
```

```python

for task_id in task_ids:
    output_dir = f"{output_root}/spatial_task_{task_id:02d}"
    command = [
        "lerobot-eval",
        f"--policy.path={policy}",
        "--env.type=libero",
        "--env.task=libero_spatial",
        f"--env.task_ids=[{task_id}]",
        "--eval.batch_size=1",
        f"--eval.n_episodes={episodes_per_task}",
        f"--output_dir={output_dir}",
    ]

    print("Running:", " ".join(command))
    subprocess.run(command, check=True)
```
